## ۲. امید ریاضی

```python
import numpy as np
import pandas as pd

outcomes = np.array([10.0, -10.0])
probs = np.array([0.55, 0.45])
ev = (outcomes * probs).sum()
var = (((outcomes - ev) ** 2) * probs).sum()
print(ev, var, round(var ** 0.5, 3))
```

```python
rng = np.random.default_rng(1)
bets = rng.choice(outcomes, size=100_000, p=probs)
print(round(bets.mean(), 3), round(bets.std(ddof=1), 3))

fee = 0.5
print(ev - fee, round((bets - fee).mean(), 3))
```

## ۳. بازده روزانه به‌عنوان متغیر تصادفی

```python
URL = "https://raw.githubusercontent.com/quantacademyir/quant-math/main/data/btc_daily_5y.csv"
close = pd.read_csv(URL, parse_dates=["date"], index_col="date")["close"]
r = close.pct_change().dropna()
print(len(r), f"{r.mean():.5f}", f"{r.std():.4f}", f"{(r > 0).mean():.1%}")
print(f"{r.min():.3f} {r.max():.3f}")
```

## ۴. خطی بودن امید

```python
x = r.values
y = 2 * x + rng.normal(0, 0.01, size=len(x))
a, b = 0.6, 0.4
print(round((a * x + b * y).mean(), 6), round(a * x.mean() + b * y.mean(), 6))
```

```python
v_sum = np.var(x + y, ddof=1)
print(round(v_sum, 6), round(np.var(x, ddof=1) + np.var(y, ddof=1), 6), round(np.var(x, ddof=1) + np.var(y, ddof=1) + 2 * np.cov(x, y)[0, 1], 6))
```

## ۵. چرا واریانس مهم است: ورشکستگی

```python
def simulate(f, p=0.55, rounds=200, paths=5000, seed=7):
    rng = np.random.default_rng(seed)
    wins = rng.random((paths, rounds)) < p
    growth = np.where(wins, 1 + f, 1 - f)
    return growth.prod(axis=1)

print(f"{'f':>5} {'theory_mean':>14} {'sim_mean':>10} {'median':>8} {'below_half':>11}")
for f in [0.02, 0.10, 0.20, 0.50, 1.00]:
    w = simulate(f)
    theory = (1 + f * 0.1) ** 200
    print(f"{f:>5.2f} {theory:>14,.1f} {w.mean():>10.2f} {np.median(w):>8.3f} {(w < 0.5).mean():>11.1%}")
```

## ۶. امید یک تابع، تابع امید نیست

```python
g = np.array([1.5, 0.5])
print(g.mean(), round(np.sqrt(g.prod()), 4))
rng = np.random.default_rng(3)
paths = rng.choice(g, size=(100000, 20))
print(round(paths.prod(axis=1).mean(), 3), round(np.median(paths.prod(axis=1)), 5))
```
